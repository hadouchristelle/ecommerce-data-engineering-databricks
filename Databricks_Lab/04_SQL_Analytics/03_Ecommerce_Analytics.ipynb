```python
########DATA ANALYTICS
# #####SILVER -> GOLD
df_silver = spark.table(
    "training_catalog.silver.ecommerce_clean"
)

df_silver.createOrReplaceTempView("ecommerce_silver_view")
```

```python
%sql
/* #GOLD TABLE 1: SALES BY COUNTRY
#Purpose:
# -- Calculate key sales metrics for each customer country.*/

CREATE OR REPLACE TABLE training_catalog.gold.sales_by_country
USING DELTA
AS

SELECT
    customer_country,

    COUNT(*) AS total_records,

    COUNT(DISTINCT order_id) AS total_orders,

    COUNT(DISTINCT customer_id) AS total_customers,

    SUM(sales_per_order) AS total_sales,

    AVG(sales_per_order) AS average_sales,

    SUM(profit_per_order) AS total_profit

FROM ecommerce_silver_view

GROUP BY customer_country;
```

```python
%sql

SELECT *
FROM training_catalog.gold.sales_by_country
ORDER BY total_sales DESC;
```

```python
%sql

CREATE OR REPLACE TABLE training_catalog.gold.sales_by_category
USING DELTA
AS

SELECT
    category_name,

    COUNT(DISTINCT order_id) AS total_orders,

    SUM(order_quantity) AS total_quantity,

    SUM(sales_per_order) AS total_sales,

    AVG(sales_per_order) AS average_sales,

    SUM(profit_per_order) AS total_profit

FROM ecommerce_silver_view

GROUP BY category_name;
```

```python
%sql

SELECT *
FROM training_catalog.gold.sales_by_category
ORDER BY total_sales DESC;
```

```python
%sql

CREATE OR REPLACE TABLE training_catalog.gold.customer_summary
USING DELTA
AS

SELECT
    customer_id,
    customer_first_name,
    customer_last_name,
    customer_country,
    customer_segment,

    COUNT(DISTINCT order_id) AS total_orders,

    SUM(order_quantity) AS total_quantity,

    SUM(sales_per_order) AS total_sales,

    AVG(sales_per_order) AS average_order_value,

    SUM(profit_per_order) AS total_profit

FROM ecommerce_silver_view

GROUP BY
    customer_id,
    customer_first_name,
    customer_last_name,
    customer_country,
    customer_segment;
```

```python
%sql

SELECT *
FROM training_catalog.gold.customer_summary
ORDER BY total_sales DESC;
```
